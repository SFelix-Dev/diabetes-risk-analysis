```python
import pandas as pd
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier

df = pd.read_csv('diabetes_012_health_indicators_BRFSS2015.csv')

df

```

```python
df = df.drop_duplicates(keep='first')
df.shape
```

------------------

**A- Define the prediction task:**

Outcome variable: diabetes_012: Whether or not the individual has diabetes or pre diabetes

2= diabetic

1= Pre diabetic

0= individual does not have diabetes



**B- Predictor variables: these may help predict the outcome because...:**

1. Age: Younger individuals may not populate the diabetic column as much as older individuals would.

2. Smoker: Smoking generally isn't good for health, and according to the FDA, it increases insulin resistance.
 
3. BMI: A higher BMI raises the risk for diabetes and also increases insulin resistance.

4. PhysActivity: Physical activity betters health, and those who aren't physically active are at risk of being diabetic.
 
5. HighBP: High blood pressure increases insulin resistance, damages blood vessels, and increases risk for organ damage.

6. HighChol: high cholesterol and diabetes normally occur together and it increase the risk for diabetes.

-----------------------------------------

**Data Preparation**

```python
#Convert one categorical feature into numeric using one hot encoding:
```

```python
categorical_columns = ['Age']

diabetes_encoded = pd.get_dummies(df, columns=categorical_columns, dtype=float, drop_first=True)

diabetes_encoded.head()
```

```python
#Check if there are missing values:
diabetes_encoded.isnull().sum()
```

```python
#No missing values found above
```

```python
#Split data into training and testing sets
age_dummy_cols = [col for col in diabetes_encoded.columns if col.startswith('Age_')]
predictors = ['Smoker', 'BMI', 'PhysActivity', 'HighBP', 'HighChol'] + age_dummy_cols

X_diabetes = diabetes_encoded[predictors]
y_diabetes = diabetes_encoded['Diabetes_012']
```

```python
X_diabetes
```

```python
from sklearn.model_selection import train_test_split

X_train_diabetes, X_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(X_diabetes, y_diabetes, test_size=0.2, random_state=42)
```

```python
#Balance the training data
y_train_diabetes.value_counts()
```

The data set is imbalanced because the number of individuals without diabetes (0) is much higher than the number of individuals with diabetes (2) and prediabetes (1). The model may favor the largest group and have trouble predicting the smaller groups, so the training data should be balanced.

```python
#balance
from imblearn.over_sampling import RandomOverSampler

randOverSample = RandomOverSampler(random_state=42)
X_train_diabetes_balanced, y_train_diabetes_balanced = randOverSample.fit_resample(X_train_diabetes, y_train_diabetes)
```

---------

**Model Selection with 5-fold Cross Validation**

**Initialize Logistic Regression and train a logistic regression model using 5-fold cross-validation:**

```python
logreg = LogisticRegression(max_iter=2000)

accuracy_scores_logreg = cross_val_score(logreg, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='accuracy')
print("Logistic Regression - Accuracy Scores across folds:", accuracy_scores_logreg)
print("Logistic Regression - Mean Accuracy:", accuracy_scores_logreg.mean())
print(" ")

precision_scores_logreg = cross_val_score(logreg, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='precision_weighted')
print("Logistic Regression - Precision Scores across folds:", precision_scores_logreg)
print("Logistic Regression - Mean Precision:", precision_scores_logreg.mean())
print(" ")

recall_scores_logreg = cross_val_score(logreg, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='recall_weighted')
print("Logistic Regression - Recall Scores across folds:", recall_scores_logreg)
print("Logistic Regression - Mean Recall:", recall_scores_logreg.mean())
print(" ")

f1_scores_logreg = cross_val_score(logreg, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='f1_weighted')
print("Logistic Regression - F1 Scores across folds:", f1_scores_logreg)
print("Logistic Regression - Mean F1 Score:", f1_scores_logreg.mean())
```

-----

 **Initialize Decision tree and train it using 5-fold cross-validation:**

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier(max_depth=10)

accuracy_scores_tree = cross_val_score(tree, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='accuracy', n_jobs=-1)
print("Decision Tree - Accuracy Scores across folds:", accuracy_scores_tree)
print("Decision Tree - Mean Accuracy:", accuracy_scores_tree.mean())
print(" ")

precision_scores_tree = cross_val_score(tree, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='precision_weighted', n_jobs=-1)
print("Decision Tree - Precision Scores across folds:", precision_scores_tree)
print("Decision Tree - Mean Precision:", precision_scores_tree.mean())
print(" ")

recall_scores_tree = cross_val_score(tree, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='recall_weighted', n_jobs=-1)
print("Decision Tree - Recall Scores across folds:", recall_scores_tree)
print("Decision Tree - Mean Recall:", recall_scores_tree.mean())
print(" ")

f1_scores_tree = cross_val_score(tree, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='f1_weighted', n_jobs=-1)
print("Decision Tree - F1 Scores across folds:", f1_scores_tree)
print("Decision Tree - Mean F1 Score:", f1_scores_tree.mean())
```

----

**Initialize random forest and train it using 5-fold cross-validation**

```python
rf = RandomForestClassifier(n_estimators=100)

accuracy_scores_rf = cross_val_score(rf, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='accuracy')
print("Random Forest - Accuracy Scores across folds:", accuracy_scores_rf)
print("Random Forest - Mean Accuracy:", accuracy_scores_rf.mean())
print(" ")

precision_scores_rf = cross_val_score(rf, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='precision_weighted')
print("Random Forest - Precision Scores across folds:", precision_scores_rf)
print("Random Forest - Mean Precision:", precision_scores_rf.mean())
print(" ")

recall_scores_rf = cross_val_score(rf, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='recall_weighted')
print("Random Forest - Recall Scores across folds:", recall_scores_rf)
print("Random Forest - Mean Recall:", recall_scores_rf.mean())
print(" ")

f1_scores_rf = cross_val_score(rf, X_train_diabetes_balanced, y_train_diabetes_balanced, cv=5, scoring='f1_weighted')
print("Random Forest - F1 Scores across folds:", f1_scores_rf)
print("Random Forest - Mean F1 Score:", f1_scores_rf.mean())
```

------

**Final Model:**

**Retrain the best model on the whole training set and test it. Report accuracy, F1, Recall, and precision.**

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

rf.fit(X_train_diabetes_balanced, y_train_diabetes_balanced)
y_pred_diabetes = rf.predict(X_test_diabetes)

test_accuracy = accuracy_score(y_test_diabetes, y_pred_diabetes)
test_precision = precision_score(y_test_diabetes, y_pred_diabetes, average='weighted')
test_recall = recall_score(y_test_diabetes, y_pred_diabetes, average='weighted')
test_f1 = f1_score(y_test_diabetes, y_pred_diabetes, average='weighted')

print("Test Accuracy (Best Model):" + str(test_accuracy))
print("Test Precision (Best Model):" + str(test_precision))
print("Test Recall (Best Model):" + str(test_recall))
print("Test F1 Score (Best Model):" + str(test_f1))
```

Interpretation: The final Random Forest model performed well, achieving a test accuracy of 0.540, a precision of 0.812(a noticeable jump from 0.595, the initializing training model), a recall of 0.540, and an F1 score of 0.630. All RF performance scores are slightly higher than those of the other two models implemented above. Also, an accuracy of 0.54 means the model gets the correct category right about half the time, which is better than random guessing. Finally, I chose Random Forest as the best-performing model because it achieved the highest mean accuracy, precision, recall, and F1 across all 5 cross-validation folds, compared to Logistic Regression and Decision Tree.

**Logistic Regression with L1 Regularization**

```python
# Logistic Regression with Lasso (L1) regularization
# Used to make the table easy to read: https://docs.python.org/3/library/functions.html#zip
from sklearn.metrics import classification_report

lasso_model = LogisticRegression(penalty='l1', solver='liblinear', C=0.01, max_iter=1000)
lasso_model.fit(X_train_diabetes_balanced, y_train_diabetes_balanced)
print(lasso_model.coef_)
print(list(zip(X_diabetes.columns, lasso_model.coef_[0])))

lasso_pred = lasso_model.predict(X_test_diabetes)
print(" ")
print("Lasso (L1) Regularization Test Set Evaluation:")
print(classification_report(y_test_diabetes, lasso_pred))
```

Interpretation: Compared to the best-performing model from 5-fold cross-validation, L1 performs better across all metrics. Random Forest, test set scores: accuracy 0.540, precision 0.812, recall 0.540, F1 0.630, where L1 scores an accuracy of 0.64, a weighted precision of 0.82, a weighted recall of 0.64, and a weighted F1 of 0.70. The variables that zeroed out were Age_4.0 and Age_5.0 (not diabetic), Smoker (pre-diabetic), and Age_6.0 (diabetic), and their removal makes sense to me because that information is likely the same or close to BMI, HighBP, and HighChol, since the rest of the variables already cover what those groups tell us.

------

**Logistic regression with L2 Regularization**

```python
# Logistic Regression with Ridge (L2) regularization
ridge_model = LogisticRegression(penalty='l2', solver='liblinear')
ridge_model.fit(X_train_diabetes_balanced, y_train_diabetes_balanced)


print(ridge_model.coef_)
print(list(zip(X_diabetes.columns, ridge_model.coef_[0])))
      
ridge_pred = ridge_model.predict(X_test_diabetes)
print (" ")
print(classification_report(y_test_diabetes, ridge_pred))
```

Interpretation: L2 performed almost the same as L1, scoring an accuracy of 0.63, a precision of 0.82, a recall of 0.63, and an F1 score of 0.69. No coefficients were zeroed out here, as expected, since L2 only shrinks coefficients rather than eliminating/zeroing them. Being that L1 and L2 scored nearly the same, even after L1 dropped Age_4.0, Age_5.0, Smoker, and Age_6.0, makes me think that those variables didn't add much value.

----

**Logistic Regression with Elastic Net Regularization**

```python
# Logistic Regression with Elastic Net regularization (Combination of L1 and L2)
elastic_net_model = LogisticRegression(penalty='elasticnet', C=0.05, solver='saga', l1_ratio=0.5)
elastic_net_model.fit(X_train_diabetes_balanced, y_train_diabetes_balanced)

print(elastic_net_model.coef_)
print(list(zip(X_diabetes.columns, elastic_net_model.coef_[1])))

elastic_net_pred = elastic_net_model.predict(X_test_diabetes)
print(" ")
print(classification_report(y_test_diabetes, elastic_net_pred))
```

Interpretation: Elastic Net zeroed out Smoker, PhysActivity, and HighBP in the "pre-diabetic" row, indicating that those three variables add little value for identifying pre-diabetic cases, though they're still useful for the other two classes. When it comes to the performance, it scores a precision of 0.82, a recall of 0.62, an F1 of 0.68, and an accuracy of 0.62, which are all slightly below L1 (accuracy of 0.64, a weighted precision of 0.82, a weighted recall of 0.64, and a weighted F1 of 0.70,) and L2(accuracy of 0.63, a precision of 0.82, a recall of 0.63, and an F1 score of 0.69). Compared to the best model (random forest: accuracy 0.540, precision 0.812, recall 0.540, F1 0.630), Elastic Net scores higher across all categories.
